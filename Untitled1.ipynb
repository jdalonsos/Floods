```python
import pandas as pd

xlsx = r"D:\M2_MoSEF\Acclim\DataCollection\data\JRC_flood_depth_maps\_outputs_country_region_department\events_country_region_department_2015_2024.xlsx"
comm_long = r"D:\M2_MoSEF\Acclim\DataCollection\data\JRC_flood_depth_maps\_outputs_inseecom\events_communes_long_all_years.csv"
events_country = pd.read_excel(xlsx, sheet_name="events_country")
summary_depts  = pd.read_excel(xlsx, sheet_name="summary_event_depts")
comm = pd.read_csv(comm_long, dtype=str)

fr_nuts_events = set(events_country.loc[events_country["country_code"]=="FR","event_id"])
fr_comm_events = set(comm["event_id"].unique())

print("France events (NUTS-based):", len(fr_nuts_events))
print("France events (commune-based):", len(fr_comm_events))
print("Missing in NUTS:", len(fr_comm_events - fr_nuts_events))

# Check: those “missing” events were still processed by the NUTS script?
missing = list(fr_comm_events - fr_nuts_events)[:10]
print("Example missing:", missing)

for eid in missing:
    r = summary_depts.loc[summary_depts["event_id"]==eid, ["event_id","n_departments","n_candidates_tested","n_tiles"]]
    print(r.to_string(index=False))
    break
```
